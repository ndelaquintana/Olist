# Análisis del dataset Olist

Carga de todas las tablas del dataset brasileño de e-commerce de Olist usando pandas.

Las tablas se encuentran en la carpeta `data/`.

```python
import pandas as pd
from pathlib import Path

pd.set_option("display.max_columns", None)

DATA_DIR = Path("../Data")  # la carpeta Data está al mismo nivel que Notebooks
```

## Carga de todas las tablas

Cargamos cada CSV en su propio DataFrame y, además, en un diccionario `tablas` para acceso programático.

```python
# Mapeo de nombre corto -> archivo CSV
archivos = {
    "customers": "olist_customers_dataset.csv",
    "geolocation": "olist_geolocation_dataset.csv",
    "order_items": "olist_order_items_dataset.csv",
    "order_payments": "olist_order_payments_dataset.csv",
    "order_reviews": "olist_order_reviews_dataset.csv",
    "orders": "olist_orders_dataset.csv",
    "products": "olist_products_dataset.csv",
    "sellers": "olist_sellers_dataset.csv",
    "category_translation": "product_category_name_translation.csv",
}

# Diccionario con todos los DataFrames
tablas = {nombre: pd.read_csv(DATA_DIR / archivo) for nombre, archivo in archivos.items()}

# Variables individuales para comodidad
customers = tablas["customers"]
geolocation = tablas["geolocation"]
order_items = tablas["order_items"]
order_payments = tablas["order_payments"]
order_reviews = tablas["order_reviews"]
orders = tablas["orders"]
products = tablas["products"]
sellers = tablas["sellers"]
category_translation = tablas["category_translation"]

print("Tablas cargadas correctamente:")
for nombre, df in tablas.items():
    print(f"  {nombre:22s} {df.shape[0]:>8,} filas x {df.shape[1]:>2} columnas")
```

## Vista rápida de cada tabla

## Unión de orders + customers + order_items

- `orders` ⋈ `customers` por `customer_id` (cada pedido tiene un cliente).
- El resultado ⋈ `order_items` por `order_id` (cada pedido puede tener varios ítems).

Usamos `how="left"` para conservar todos los pedidos.

```python
# 1) orders + customers por customer_id
orders_customers = orders.merge(customers, on="customer_id", how="left")

# 2) + order_items por order_id
orders_full = orders_customers.merge(order_items, on="order_id", how="left")

print(f"orders............: {orders.shape}")
print(f"orders+customers..: {orders_customers.shape}")
print(f"+ order_items.....: {orders_full.shape}")
print(f"\nPedidos únicos en el resultado final: {orders_full['order_id'].nunique():,}")
print(f"Filas (a nivel de ítem de pedido)...: {len(orders_full):,}")

orders_full.head()
```

## Unión simple: customers ⋈ orders

Cada pedido (`orders`) pertenece a un único cliente (`customers`), enlazados por `customer_id`.
Hacemos un `left` join partiendo de `customers` para conservar todos los clientes
(incluso los que no tuvieran ningún pedido).

```python
# customers + orders por customer_id
customers_orders = customers.merge(orders, on="customer_id", how="left")

print(f"customers.........: {customers.shape}")
print(f"orders............: {orders.shape}")
print(f"customers+orders..: {customers_orders.shape}")
print(f"\nClientes sin pedido: {customers_orders['order_id'].isna().sum():,}")

customers_orders.head()
```
